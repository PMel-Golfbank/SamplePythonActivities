```python
import math

x1 = float(input("Enter a value for x1: "))
y1 = float(input("Enter a value for y1: "))
x2 = float(input("Enter a value for x2: "))
y2 = float(input("Enter a value for y2: "))

#This line will compute the distance of the two points
distance = math.sqrt(math.pow(x2 - x1, 2) + math.pow(y2 - y1, 2))

#The purpose of round is to round off the decimal number of distance into two decimal places
print("The distance between the two points is ", str(round(distance, 2)))
```
